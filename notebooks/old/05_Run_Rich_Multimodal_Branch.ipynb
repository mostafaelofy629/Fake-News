# Run Rich Multimodal Branch

This notebook runs only the final OCR + BLIP + BERT + ResNet50 enriched fusion branch from `03_Multimodal_Model_Training_FINAL_Enhanced.ipynb`.

It uses the existing train/validation/test-safe design from Notebook 03 and writes outputs to `final_artifacts/`.

If cached rich features already exist, it runs quickly. To force full OCR/BLIP feature extraction again, set `RICH_FORCE_REBUILD_FEATURES = True` inside the rich branch cell in Notebook 03 before running this notebook.

```python
from pathlib import Path
import nbformat
from IPython.display import display

NOTEBOOK_03 = Path('03_Multimodal_Model_Training_FINAL_Enhanced.ipynb')
FINAL_ARTIFACT_DIR = Path('final_artifacts')
FINAL_ARTIFACT_DIR.mkdir(exist_ok=True)

print('Notebook 03 exists:', NOTEBOOK_03.exists())
print('Artifact directory:', FINAL_ARTIFACT_DIR.resolve())
```

```python
nb = nbformat.read(NOTEBOOK_03, as_version=4)
required_markers = [
    '# FINAL UPGRADE 03-M1',
    '# FINAL UPGRADE 03-M2',
    '# FINAL UPGRADE 03-M3',
    '# FINAL UPGRADE 03-M4',
    '# FINAL UPGRADE 03-M7',
    '# FINAL UPGRADE 03-M6',
]

cells_to_run = []
for marker in required_markers:
    for idx, cell in enumerate(nb.cells):
        if cell.cell_type == 'code' and marker in cell.source:
            cells_to_run.append((idx, marker, cell.source))
            break
    else:
        raise RuntimeError(f'Could not find required cell marker: {marker}')

for idx, marker, _ in cells_to_run:
    print(f'Found {marker} at Notebook 03 cell {idx}')
```

```python
# Run the selected Notebook 03 cells in order.
# This keeps Notebook 03 as the source of truth while giving you an .ipynb runner.

namespace = globals()
for idx, marker, source in cells_to_run:
    print(f'\n===== RUNNING Notebook 03 cell {idx}: {marker} =====')
    exec(compile(source, f'{NOTEBOOK_03.name}:cell-{idx}', 'exec'), namespace)
    print(f'===== FINISHED Notebook 03 cell {idx}: {marker} =====')

print('\nRich multimodal notebook run finished.')
```
