# Build Content + Title Modeling Split Folders

This notebook creates two modeling split folders:

1. `data_splits/content_title_valid_image_multimodal/`
   - only rows with valid images
   - for multimodal models

2. `data_splits/content_title_text_only_all/`
   - all rows, with or without images
   - for text-only models

Both folders include `title`, `content`, and `text_for_model`. The `text_for_model` column uses article body `content` when available and falls back to `title` when content extraction failed or was not attempted.

```python
from pathlib import Path
import pandas as pd

print('Builder exists:', Path('build_content_title_modeling_splits.py').exists())
```

```python
import build_content_title_modeling_splits as builder

summary = builder.build()
display(summary)
```

```python
for folder in [
    Path('data_splits/content_title_valid_image_multimodal'),
    Path('data_splits/content_title_text_only_all'),
]:
    print('\n', folder)
    for split in ['train', 'val', 'test']:
        path = folder / f'{split}.csv'
        df = pd.read_csv(path)
        print(split, 'rows=', len(df), 'content_available=', int(df['content_available'].sum()))
        display(df[['id', 'label', 'title', 'content_available', 'text_source_for_model']].head(3))
```
